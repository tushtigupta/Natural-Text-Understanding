```python
from langchain_groq import ChatGroq
from langchain_core.prompts import PromptTemplate
import os
```

```python
os.environ["GROQ_API_KEY"] = "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"
```

```python
llm = ChatGroq(
    api_key=os.environ["GROQ_API_KEY"],
    model_name="llama-3.3-70b-versatile",
    temperature=0
)
```

```python
template = """
Create a SQL query snippet using the below text:

{text}

Only return the SQL query.
"""

prompt = PromptTemplate(
    template=template,
    input_variables=["text"]
)
```

```python
llm_chain = prompt | llm
```

```python
text = """
extract name and salary columns
"""
```

## Generate SQL Query

```python
response = llm_chain.invoke({"text": text})

print(response.content)
```

## More Example

```python
text = "Find all employees whose salary is smaller than400000"

response = llm_chain.invoke({"text":
                             text})

print(response.content)
```
