```python
 import fitz

doc = fitz.open("/content/UKSim2014_IEEE.pdf")

text = ""

for page in doc:
    text += page.get_text()

#print(text[:1000])***
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200
)

chunks = splitter.split_text(text)

print(len(chunks))
```
