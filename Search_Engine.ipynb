```python
from langchain_community.vectorstores import FAISS
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_core.documents import Document
from langchain_groq import ChatGroq
```

```python
docs = [
    Document(page_content="LSTM is a recurrent neural network."),

    Document(page_content="GRU is faster than LSTM."),

    Document(page_content="Transformers are powerful NLP models."),

    Document(page_content="Vector embeddings enable semantic search."),

    Document(page_content="Python is popular in AI and ML.")
]
```

```python
embedding_model = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
vectorstore = FAISS.from_documents(
    docs,
    embedding_model
)
```

```python
print("Vector DB Created Successfully")
```

```python
llm = ChatGroq(
    model_name="llama3-70b-8192"
)
```

```python
def semantic_search(query):

    print(f"\nQuery: {query}\n")

    # Similarity Search
    results = vectorstore.similarity_search(query, k=2)

    context = "\n".join([doc.page_content for doc in results])

    print("Retrieved Documents:\n")

    for r in results:
        print("-", r.page_content)

    # Prompt
    prompt = f"""
    Answer the question using the context below.

    Context:
    {context}

    Question:
    {query}
    """

    # LLM Response
    response = llm.invoke(prompt)

    print("\nLLM Answer:\n")
    print(response.content)
```

```python
semantic_search("Which model is used in NLP?")
semantic_search("What helps semantic search?")
semantic_search("Tell me about recurrent networks")
```
